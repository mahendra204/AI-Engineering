```python
spark = SparkSession.builder \
    .appName("MyApp") \
    .master("local[1]") \
    .config("spark.sql.shuffle.partitions", "1") \
    .config("spark.default.parallelism", "1") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")
```

```python
df = spark.createDataFrame([(1, "Alice"), (2, "Bob"), (3, "Charlie")], schema=["id", "name"])
```

```python
df.printSchema()
```

```python
import pandas as pd

# Create dataframe from the same data source
data = [(1, "Alice"), (2, "Bob"), (3, "Charlie")]
pdf = pd.DataFrame(data, columns=["id", "name"])
pdf
```

```python
# Method 1: From list of tuples (you already did this)
data1 = [(1, "Alice", 25), (2, "Bob", 30), (3, "Charlie", 35)]
df1 = spark.createDataFrame(data1, schema=["id", "name", "age"])
print("Method 1 - From list of tuples:")
df1.printSchema()

# Method 2: From list with explicit schema
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

schema = StructType([
    StructField("user_id", IntegerType(), True),
    StructField("username", StringType(), True),
    StructField("city", StringType(), True)
])

data2 = [(1, "john_doe", "New York"), (2, "jane_smith", "Los Angeles")]
df2 = spark.createDataFrame(data2, schema=schema)
print("\nMethod 2 - From list with explicit schema:")
df2.printSchema()

# Method 3: From Pandas DataFrame
import pandas as pd
pdf_data = pd.DataFrame({
    "product_id": [101, 102, 103],
    "product_name": ["Laptop", "Phone", "Tablet"],
    "price": [1000, 500, 300]
})
df3 = spark.createDataFrame(pdf_data)
print("\nMethod 3 - From Pandas DataFrame:")
df3.show()
```
